```python
df_p["tipoDificultad"].unique().tolist()
```

```python
def disability_to_dummy(x):
    if (x == "NINGUNA" or
        x == "SE IGNORA"):
        return 0
    else:
        return 1
```

```python
patient["disability"] = df_p.loc[3:, "tipoDificultad"].map(disability_to_dummy)
patient.tail(10)
```
