## LLM

```python
from dotenv import load_dotenv
load_dotenv()
llm = ChatGroq(model_name="llama3-70b-8192")
```

```python
df_hq = SmartDataframe(df_hq, config={"llm": llm})
df_hi = SmartDataframe(df_hi, config={"llm": llm})
df_np = SmartDataframe(df_np, config={"llm": llm})
```

```python
ai = Agent([df_hq, df_hi, df_np], config={"llm": llm}, memory_size=10)
```

```python
ai.chat("¿Cuántos Expedientes repetidos existen?")
```

```python
ai.chat("¿Cuántos registros coinciden en Expediente y Registro? Ponlos en una tabla")
```
